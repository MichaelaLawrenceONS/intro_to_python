## Working with Jupyter Notebooks

The training materials are notebooks**, which allow us to combine written explanations, code, and outputs in a single interactive document.

### Cells

A Jupyter notebook is made up of **cells**. Each cell is a container that can hold either text or code, and cells are executed individually.

There are two main cell types you’ll encounter:

- **Markdown cells**  
  These cells contain formatted text, such as explanations, headings, bullet points, and links. They are used to describe concepts or provide instructions and do not run any code. This cell is a markdown cell.

- **Code cells**  
  These cells contain Python code. When you run a code cell, Python executes the code and displays the output directly below the cell (for example, printed text, tables, or charts).

You can switch a cell’s type using the toolbar or keyboard shortcuts, and you can add, delete, or reorder cells as needed.

### Running Cells and Execution Order

One important thing to understand about Jupyter notebooks is that **the order in which cells are run matters**.

- Code cells do **not** have to be run from top to bottom.
- When a cell is run, any variables, functions, or imports it creates are stored in memory.
- If you try to use something that hasn’t been defined yet — or was defined in a cell that hasn’t been run — you may see errors.

Because of this, running cells out of order can sometimes cause confusing or unexpected behaviour.

### Restarting the Kernel

The **kernel** is the process that runs your Python code and keeps track of everything currently stored in memory.

If you encounter a problem such as:
- Variables behaving unexpectedly
- Errors that don’t seem to make sense
- Code that worked before but no longer does

a common and effective solution is to:

**Restart the kernel and run all cells**

You can usually do this from the menu:  
`Kernel → Restart Kernel and Run All`

## Introduction to Data Types

In the previous section, you used Python as a calculator to perform calculations like `2 + 2` or `2 * (3 + 2)`. Although it may not have been obvious at the time, you were already working with **data types**.

A **data type** describes the kind of data a value represents, and it determines what you can do with that value.

### Integers (`int`) and Floating‑Point Numbers (`float`)

- An **integer** (`int`) is a whole number, such as `1`, `10`, or `-3`.
- A **floating‑point number** (`float`) is a number with a decimal point, such as `3.14`, `0.5`, or `2.0`.

### Strings (`str`)
Another common and very important data type is the string (`str`).
A string represents text, such as words or sentences, and is created by enclosing characters in quotes:
- "Hello"
- 'Python'
- "1"

```python
"Hello" + '100'
```
